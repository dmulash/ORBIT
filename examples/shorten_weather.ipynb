```python
def filter_csv_by_date(input_csv, output_csv, date_column):
    # Load the CSV file
    df = pd.read_csv(input_csv, parse_dates=[date_column])
    
    # Filter out rows where the date is before 1976
    df_filtered = df[df[date_column] >= '1976-01-01']
    
    # Save the filtered dataframe to a new CSV file
    df_filtered.to_csv(output_csv, index=False)
    
    print(f"Filtered data saved to {output_csv}")

# Example usage
input_csv = "data/boem_era5_40.5_-73.0.csv"  # Replace with your actual file name ,  
output_csv = "data/boem_era5_40.5_-73.0.csv"
date_column = 'datetime'  # Replace with the actual column name containing dates

filter_csv_by_date(input_csv, output_csv, date_column)
```
